```python
import nltk
import string
import random

from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('stopwords')

class TextPreprocessor:

    def __init__(self):
        self.lemmatizer = WordNetLemmatizer()
        self.stop_words = set(stopwords.words("english"))

    def remove_punctuation(self, text):
        return text.translate(
            str.maketrans("", "", string.punctuation)
        )

    def tokenize(self, text):
        return word_tokenize(text)

    def lemmatize(self, tokens):
        return [
            self.lemmatizer.lemmatize(token)
            for token in tokens
        ]

    def remove_stop_words(self, tokens):
        return [
            token
            for token in tokens
            if token not in self.stop_words
        ]

    def preprocess(self, text):

        text = text.lower()

        text = self.remove_punctuation(text)

        tokens = self.tokenize(text)

        tokens = self.remove_stop_words(tokens)

        tokens = self.lemmatize(tokens)

        return " ".join(tokens)
# AI POWERED CHATBOT

class AIPoweredChatbot:

    def __init__(self):

        self.preprocessor = TextPreprocessor()

        # CountVectorizer converts text into numerical features
        self.vectorizer = CountVectorizer()

        # Naive Bayes Machine Learning model
        self.model = MultinomialNB()

        # Combine vectorizer + Naive Bayes
        self.pipeline = make_pipeline(
            self.vectorizer,
            self.model
        )

        # Train the chatbot
        self.train_data()

        self.responses = {

            "greeting": [
                "Hello! How can I assist you today?",
                "Hi there! What can I do for you?",
                "Hey! Nice to meet you!"
            ],

            "how_are_you": [
                "I'm doing well! How can I help you?",
                "I'm just a program, but I'm here to help you!"
            ],

            "name": [
                "I am an AI-powered chatbot.",
                "You can call me Chatbot."
            ],

            "help": [
                "Sure! I can help you. What do you need assistance with?",
                "Of course! Tell me what you need help with."
            ],

            "capabilities": [
                "I can answer questions, provide information and assist with simple tasks.",
                "I can chat with you and answer questions based on my training."
            ],

            "joke": [
                "Why did the computer go to the doctor? Because it had a virus!",
                "Why was the computer cold? Because it left its Windows open!"
            ],

            "weather": [
                "I don't have real-time weather information, but I can provide general weather advice."
            ],

            "password": [
                "To reset your password, please use the Forgot Password option on the website or contact support."
            ],

            "account": [
                "Please provide more details about your account issue so I can assist you better."
            ],

            "book": [
                "I recommend 'To Kill a Mockingbird' by Harper Lee. It's a classic!"
            ],

            "meaning_of_life": [
                "The meaning of life is a philosophical question that has been debated for centuries."
            ],

            "support": [
                "You can contact support through the official support page or customer service."
            ],

            "goodbye": [
                "Goodbye! Have a great day!",
                "See you later! Take care!"
            ]
        }

    def train_data(self):

        training_data ={

            "greeting": [
                "hello",
                "hi",
                "hey",
                "hello there",
                "hi there",
                "hey there",
                "good morning",
                "good afternoon",
                "good evening"
            ],

            "how_are_you": [
                "how are you",
                "how are you doing",
                "how are things",
                "are you fine",
                "how do you feel"
            ],

            "name": [
                "what is your name",
                "what's your name",
                "who are you",
                "tell me your name",
                "what should I call you"
            ],


            "help": [
                "help",
                "help me",
                "I need help",
                "I need assistance",
                "can you help me",
                "could you help me",
                "I need your help"
            ],

          "capabilities": [
                "what can you do",
                "what are your capabilities",
                "what can you help with",
                "tell me what you can do",
                "what services do you provide"
            ],

            "joke": [
                "tell me a joke",
                "say a joke",
                "make me laugh",
                "do you know any jokes",
                "give me a funny joke"
            ],

            "weather": [
                "what is the weather",
                "how is the weather",
                "weather today",
                "tell me about the weather",
                "what is the weather like"
            ],

            "password": [
                "how do I reset my password",
                "I forgot my password",
                "reset password",
                "change my password",
                "password reset"
            ],

            "account": [
                "I have a problem with my account",
                "account problem",
                "my account is not working",
                "I have an account issue",
                "help with my account"
            ],

            "book": [
                "recommend a book",
                "suggest a good book",
                "what book should I read",
                "can you recommend a book",
                "give me a book recommendation"
            ],

            "meaning_of_life": [
                "what is the meaning of life",
                "why are we alive",
                "what is life's purpose",
                "meaning of life",
                "purpose of life"
            ],

            "support": [
                "how do I contact support",
                "contact support",
                "I want to contact customer service",
                "where can I get support",
                "customer support"
            ],

            "goodbye": [
                "bye",
                "goodbye",
                "see you",
                "see you later",
                "good night",
                "quit",
                "exit"
            ]
        }

        sentences = []
        labels = []

        for intent, examples in training_data.items():

            for example in examples:

                processed_text = self.preprocessor.preprocess(
                    example
                )

                sentences.append(processed_text)

                labels.append(intent)

        self.pipeline.fit(sentences, labels)

    def get_response(self, user_input):

        # Preprocess ONLY ONCE
        processed_input = self.preprocessor.preprocess(
            user_input
        )

        # Predict the intent using Naive Bayes
        predicted_intent = self.pipeline.predict(
            [processed_input]
        )[0]

        # Select a response from the predicted intent
        response = random.choice(
            self.responses[predicted_intent]
        )

        return response

print("=" * 55)
print("             AI POWERED CHATBOT")
print("=" * 55)

ai_bot = AIPoweredChatbot()

print("\nChatbot: Hello! I am an AI-powered chatbot.")
print("Chatbot: You can ask me questions or say hi.")
print("Chatbot: Type 'exit' or 'quit' to stop.\n")
while True:

    user_input = input("User: ")

    if user_input.lower().strip() in ["exit", "quit"]:

        print("Chatbot: Goodbye! Have a great day!")

        break
    if not user_input.strip():

        print("Chatbot: Please enter something.")

        continue
    response = ai_bot.get_response(user_input)

    print("Chatbot:", response)
```
