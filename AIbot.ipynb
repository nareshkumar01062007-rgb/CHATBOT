```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk.chat.util import Chat, reflections
import sklearn
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
import json
import re
import random
import string

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('punkt_tab') # Added based on the LookupError

class TextPreprocessor:

    def __init__(self):
        self.lemmatizer = WordNetLemmatizer()
        self.stop_words = set(nltk.corpus.stopwords.words("english"))

    def remove_punctuation(self, text):
        return text.translate(str.maketrans("", "", string.punctuation))

    def tokenize(self, text):
        return word_tokenize(text)

    def lemmatize(self, tokens):
        return [
            self.lemmatizer.lemmatize(token)
            for token in tokens
        ]

    def remove_stop_words(self, tokens):
        return [
            token for token in tokens
            if token not in self.stop_words
        ]

    def preprocess(self, text):
        text = text.lower()
        text = self.remove_punctuation(text)
        tokens = self.tokenize(text)
        tokens = self.remove_stop_words(tokens)
        tokens = self.lemmatize(tokens)
        return " ".join(tokens)
class AIPoweredChatbot:

    def __init__(self):
        self.chatbot = Chat(pairs, reflections)
        self.vectorizer = CountVectorizer()
        self.model= MultinomialNB()
        self.pipeline = make_pipeline(self.vectorizer, self.model)
        self.train_data()
        self.preprocessor = TextPreprocessor()
    def train_data(self):
        # Sample training data
        training_data = [
            "hello","hi", "hey", "how are you", "what is your name", "help", "bye",
            "goodbye", "I need assistance", "Can you help me?", "What can you do?", "Tell me a joke",
            "What is the weather like?", "How do I reset my password?", "I have a question about my account",
            "Can you recommend a good book?", "What is the meaning of life?", "How do I contact support?"
        ]
        responses=[
            "Hello! How can I assist you today?",
            "Hi there! What can I do for you?",
            "Hey! How's it going?",
            "I'm just a program, but I'm here to help you!",
            "I am a chatbot created to assist you.",
            "Sure, I can help you. What do you need assistance with?",
            "Goodbye! Have a great day!",
            "See you later! Take care!",
            "I'm here to assist you with any questions or concerns.",
            "Of course! What do you need help with?",
            "I can provide information, answer questions, and assist with various tasks.",
            "Why did the scarecrow win an award? Because he was outstanding in his field!",
            "I don't have real-time weather information, but I can provide general weather advice.",
            "To reset your password, please follow the instructions on our website or contact support.",
            "Please provide more details about your account issue so I can assist you better.",
            "I recommend 'To Kill a Mockingbird' by Harper Lee. It's a classic!",
            "The meaning of life is a philosophical question that has been debated for centuries.",
            "You can contact support through our website or by calling our customer service number."
        ]
        self.pipeline.fit(training_data, responses)

    def get_response(self, user_input):
        preprocessed_input = self.preprocessor.preprocess(user_input)
        return self.chatbot.respond(preprocessed_input)
def chatbot_response(user_input):
    user_input=user_input.lower()
    if "hello" in user_input or "hi" in user_input:
        return "Hello! How can I assist you today?"
    elif "how are you" in user_input:
        return "I'm just a program, but I'm here to help you!"
    elif "your name" in user_input:
        return "I am a chatbot created to assist you."
    elif "bye" in user_input or "goodbye" in user_input:
        return "Goodbye! Have a great day!"
    else:
        return "I'm sorry, I didn't understand that. Can you please rephrase?"
pairs=[
    [r"hi|hello|hey",
     ["Hello! How can I assist you today?","Hi there! What can I do for you?","Hey! How's it going?"]],
    [r"how are you ?",
     ["I'm just a program, but I'm here to help you!","I'm doing well, thank you! How can I assist you?"]],
    [r"what is your name ?",
     ["I am a chatbot created to assist you.","You can call me Chatbot."]],
     [r"(.*) help (.*)",
     ["Sure, I can help you. What do you need assistance with?","I'm here. How can I help you?"]],
     [r"bye|goodbye",
     ["Goodbye! Have a great day!","See you later! Take care!"]],
     [r"(.*)",
     ["I'm sorry, I didn't understand that. Can you please rephrase?","Can you please clarify your question?."]]
]
chatbot=Chat(pairs, reflections)


print("Basic Rule-Based Chatbot Initialized.")
processor = TextPreprocessor()
ai_bot = AIPoweredChatbot()
print("/n")
print("Chatbot: Hello! I am a simple chatbot. You can ask me questions or say hi.")
while True:
    user_input = input("User: ")

    preprocessed_text = processor.preprocess(user_input)
    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Goodbye! Have a great day!")
        break

    response = ai_bot.get_response(preprocessed_text)
    print(f"Chatbot: {response}")
```
